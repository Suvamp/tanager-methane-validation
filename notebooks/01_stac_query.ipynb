## Explore the Catalog Structure

```python
import pystac
import requests

# Correct URL — append /catalog.json to get the actual JSON
catalog = pystac.Catalog.from_file(
    "https://www.planet.com/data/stac/catalog.json"
)

print(f"Catalog title: {catalog.title}")
print("\nAvailable collections/catalogs:")
for child in catalog.get_children():
    print(f"  - {child.id}: {child.title}")
```

## List all GHG Plume scenes

```python
import pystac

ghg_collection = pystac.Collection.from_file(
    "https://www.planet.com/data/stac/tanager-core-imagery/GHG-plumes/collection.json"
)

print(f"Collection: {ghg_collection.title}")
print(f"Temporal extent: {ghg_collection.extent.temporal.intervals}")
print(f"\nScenes ({len(list(ghg_collection.get_items()))} total):\n")

for item in ghg_collection.get_items():
    print(f"ID:      {item.id}")
    print(f"Date:    {item.datetime}")
    print(f"BBox:    {item.bbox}")
    h5_url = item.assets.get("basic_radiance_hdf5")
    print(f"HDF5:    {h5_url.href if h5_url else 'N/A'}")
```

## Download a scene's HDF5 file

```python
import requests, os

H5_URL = "https://storage.googleapis.com/open-cogs/planet-stac/tanager1-release2-core-imagery/basic_radiance_hdf5/20250205_175845_00_4001_basic_radiance_hdf5.h5"

os.makedirs("../data/raw", exist_ok=True)
out_path = "../data/raw/20250205_175845_00_4001_basic_radiance.h5"

if not os.path.exists(out_path):
    print("Downloading Permian Basin scene...")
    r = requests.get(H5_URL, stream=True)
    r.raise_for_status()
    with open(out_path, "wb") as f:
        for chunk in r.iter_content(chunk_size=8192):
            f.write(chunk)
    print(f"Done! ({os.path.getsize(out_path)/1e6:.1f} MB)")
else:
    print("Already downloaded.")
```
